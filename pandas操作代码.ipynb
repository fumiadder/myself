### pandas模块：
可以基于该模块进行excel表格的数据处理

```python
#pip install pandas
import pandas as pd
```

```python
#1.读取加载Excel文件数据
data = pd.read_excel('篮球赛.xlsx',sheet_name='aaa')
data
```

```python
#2.查看数据表格的形状
data.shape  #数据表格有28行10列
```

```python
#3.数据去重
data['胜负'].unique() #将data数据表格中胜负这一列的数据取出，调用了unique函数（数据去重）
```

```python
#4.统计元素出现的个数
data['胜负'].value_counts() #统计胜负列中每个元素出现的次数
```

```python
#5.检测数据中是否有重复的行数据
data.duplicated() #duplicated函数会针对data的每一行进行重复检测，如果发现某行重复了就给改行返回一个True，否则返回一个False
```

```python
data.duplicated().sum() #统计重复行的个数
```

```python
#6.删除重复的行数据
data.drop_duplicates()
```

```python
#7.删除空值对应的行数据
data.dropna()
```

```python
#8.可以将处理后的数据写回到Excel文件中保存
data.to_excel('./ret.xlsx')
```
